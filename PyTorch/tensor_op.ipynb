```python
import torch
```

```python
my_torch = torch.arange(10)
my_torch
```

### Reshape and View

```python
my_torch = my_torch.reshape(2,5)
my_torch
```

# We can do reshape if we dont know the exact number of items

```python
my_torch2 = torch.arange(15)
my_torch2
```

```python
my_torch2 = my_torch2.reshape(3, -1)
my_torch2
```

```python
my_torch3 = torch.arange(10)
my_torch4 = my_torch3.view(2,5)
my_torch4
```

```python
#! Difference between view and reshape is that view is need contiguous tensor, reshape doing the shaping by creating a copy, it doesnt need a contiguous tensor
```

```python
# with reshape adn view they are updates
my_torch5 = torch.arange(10)
my_torch5
```

```python
my_torch6 = my_torch5.reshape(2, 5)
my_torch6
```

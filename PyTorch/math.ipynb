```python
import torch
import numpy as np
```

```python
tensor_a = torch.tensor([1,2,3,4])
tensor_b = torch.tensor([5,6,7,8])
```

```python
# Addition 
tensor_a + tensor_b
```

```python
# Addition Longhand
torch.add(tensor_a, tensor_b)
```

```python
# Subtraction 
tensor_b - tensor_a
```

```python
# Subtraction Longhand
torch.sub(tensor_b, tensor_a)
```

```python
# Multiplication 
tensor_a * tensor_b
```

```python
# Multiplication Longhand
torch.mul(tensor_b, tensor_a)
```

```python
# Division 
tensor_b / tensor_a
```

```python
# Division Longhand
torch.div(tensor_b, tensor_a)
```

```python
# Remainders Modulus
tensor_b % tensor_a
```

```python
# Remainders Modulus Longhand
torch.remainder(tensor_b, tensor_a)
```

```python
# Exponents / power
tensor_a ** tensor_b
```

```python
# Exponents/Power Longhand
torch.pow(tensor_a, tensor_b)
```

```python
# Another way to write the longhand
tensor_a.add(tensor_b)
```

```python
# Reassignment_
tensor_a.add_(tensor_b)
tensor_a
```

```python
import numpy as np

x = np.array([1,2,3])
y = np.array([2,4,5])
N = len(x)

w = 0
b = 0
learning_rate = 0.1

def initial_loss(w, b, x, y): 
    z = w*x + b      
    return np.mean((z - y) ** 2)

in_loss = initial_loss(w,b,x,y)

print(f'initial loss = {in_loss}')

def grad_analytic(w, b, x, y): 
    y_hat = w*x + b
 
    dL_dyhat = 2 * (y_hat - y) 
    dyhat_dw = x   
    dyhat_db = 1.0   
 
    dL_dw = (1 / N) * np.sum(dL_dyhat * dyhat_dw) 
    dL_db = (1 / N) * np.sum(dL_dyhat * dyhat_db) 
 
    return dL_dw, dL_db

dL_dw, dL_db = grad_analytic(w,b,x,y)
print(f"dL_dw (analytic)= {dL_dw}")
print(f"dL_db (analytic)= {dL_db}")

def numeric_gradient(w, b, x, y, h=1e-5):
    L_w_plus = initial_loss(w + h, b, x, y)
    L_w_minus = initial_loss(w - h, b, x, y)
    dL_dw = (L_w_plus - L_w_minus) / (2 * h)
    
    L_b_plus = initial_loss(w, b + h, x, y)
    L_b_minus = initial_loss(w, b - h, x, y)
    dL_db = (L_b_plus - L_b_minus) / (2 * h)
        
    return dL_dw, dL_db

dL_dw, dL_db = numeric_gradient(w,b,x,y)
print(f"dL_dw (numeric)= {dL_dw}")
print(f"dL_db (numeric)= {dL_db}")

w = w - learning_rate * dL_dw
b = b - learning_rate * dL_db

print(f"Updated w = {w}")
print(f"Updated b = {b}")

new_loss = initial_loss(w, b, x, y)
print(f"New loss = {new_loss}")

loss_decreased = new_loss < in_loss
assert loss_decreased
print("Loss is decreased")
```

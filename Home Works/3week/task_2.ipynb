Task 2 – Choose and compare losses

A. Regression – targets [1, 2, 4], predictions [0, 2, 7].

Compute residuals and both losses; repeat with the last prediction changed to 10.
Explain the difference in sensitivity to the larger error – the two losses use different
scales, so a smaller numeric value doesn’t automatically mean “better.

```python
import numpy as np
y = np.array([1, 2, 4])
y_pred = np.array([0, 2, 7])

def MSE(y, y_pred):
    return np.mean((y-y_pred)**2)

def MAE(y, y_pred):
    return np.mean(abs(y - y_pred))

print(f'MSE = {MSE(y, y_pred)}')
print(f'MAE = {MAE(y, y_pred)}')
```

B. Binary classification – targets [1, 0]. Model A predicts [0.8, 0.2]; Model B predicts [0.6, 0.4] (probability of class 1).

For each model: mean BCE, predicted classes at threshold p > 0.5, and accuracy.
Explain how the accuracies can agree while the BCE values differ. Don’t feed thresholded classes into BCE.

```python
y = np.array([1, 0])
a_p = np.array([0.8, 0.2])
b_p = np.array([0.6, 0.4])

def BSE(y, pred):
    return -np.mean(y*np.log(pred) + (1-y)*np.log(1-pred))

print(f'Model A bce = {BSE(y, a_p)}')
print(f'Model B bce = {BSE(y, b_p)}')

pred_a = (a_p>0.5).astype(int)
pred_b = (b_p>0.5).astype(int)

print("Classes A:", pred_a)
print("Classes B:", pred_b)

print("Accuracy A:", np.mean(pred_a == y))
print("Accuracy B:", np.mean(pred_b == y))
```

C. Multiclass – logits [0, ln 3, 0], correct class index 1 (zero-based).

Compute the softmax probabilities and the loss; verify they sum to 1. Then, one
sentence each: an output representation + loss for (a) an unrestricted real value, (b)
a yes/no outcome, (c) exactly one of three categories.

```python
logits = np.array([0, np.log(3), 0])
index = 1

exp_l = np.exp(logits)
probobility = exp_l/np.sum(exp_l)

print(probobility)
assert np.sum(probobility) == 1

loss = -np.log(probobility[index])
print(loss)
```

```python
#1
import numpy as np 
import matplotlib.pyplot as plt

n1 = 0.05
n2 = 0.09
w = [2,1]
g = [2*w[0], 20*w[1]]
print(f"First two updates with 0.05:")
for i in range(2):
    w_new_1 = [w[0] - n1*g[0], w[1] - n1*g[1]]
    w = w_new_1
    g = [2*w[0], 20*w[1]]
    print(w_new_1)

print(f"Second one update with 0.09:")
w = [2,1]
g = [2*w[0], 20*w[1]]
w_new_2 = [w[0] - n2*g[0], w[1] - n2*g[1]]
print(w_new_2)
```

```python
#2
w = [2,1]
w1, w2 = np.meshgrid(np.linspace(-2.5, 2.5, 200), np.linspace(-1.5, 1.5, 200))
plt.contour(w1, w2, w1**2 + 10 * w2**2, levels=15, colors="lightgray")
plt.plot(w[0], w[1], "o")
plt.plot(w_new_1[0], w_new_1[1], "o")
plt.plot(0, 0, "o")
plt.annotate('', xy=w_new_1, xytext=w, arrowprops={"arrowstyle":"->"})
plt.annotate('', xy=(0,0), xytext=w, arrowprops={"arrowstyle":"->"})

plt.xlim(-2.5, 2.5); 
plt.ylim(-1.5, 1.5); 
plt.show()
```

```python
w = [2,1]
n = [0.01, 0.05, 0.09, 0.11]

def calculating(w, n):
    path = [w]
    for i in range(30):
        g = [2*w[0], 20*w[1]]
        w_new = [w[0] - n*g[0], w[1] - n*g[1]]
        w = w_new
        g = [2*w[0], 20*w[1]]
        path.append(w_new)
    path = np.array(path)
    return path
        
fig, axes = plt.subplots(1, 4, figsize=(18,4))

for ax, lr in zip(axes, n):
    path = calculating(w, lr)
    ax.contour(w1, w2, w1**2 + 10 * w2**2, levels=15, colors="lightgray")
    ax.plot(path[:, 0], path[:, 1], "o-", markersize=3)
    
```

Task 1 – Trace a small MLP

```python
import numpy as np
X = np.array([[1., 0.], [0., 1.], [1., 1.]])
y = np.array([[1.], [2.], [1.]])
W1 = np.array([[1., -1.], [-1., 1.]])
b1 = np.array([0.5, 0.5])
W2 = np.array([[1.], [2.]])
b2 = np.array([0.])
```

```python
def ReLU(Z1):
    return np.maximum(0, Z1)

Z1 = X @ W1 + b1
H = ReLU(Z1)
y_hat = H @ W2 + b2

assert y_hat.shape == y.shape

loss = np.mean((y - y_hat)**2)

print(loss)
```

```python
y_hat = X @ (W1 @ W2) + (b1 @ W2 + b2)

assert y_hat.shape == y.shape

loss = np.mean((y - y_hat)**2)

print(loss)
```

Task 1 — The Numeric Derivative 

```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
def f(x):
    return np.sin(x) 
 
 
def df_true(x): 
    return np.cos(x) 
 
 
def deriv_forward(x, h): 
    return (f(x+h) - f(x)) / h
 
 
def deriv_central(x, h):
    return (f(x+h) - f(x-h)) / (2*h)
```

```python
x0 = 1.0 
 
print("true           :", df_true(x0)) 
print("forward  h=1e-5:", deriv_forward(x0, 1e-5)) 
print("central  h=1e-5:", deriv_central(x0, 1e-5)) 
 
hs = np.logspace(-1, -12, 45) 
 
err_fwd = np.array([ 
    abs(deriv_forward(x0, h) - df_true(x0)) 
    for h in hs 
]) 
print(f"Array of steps -> {hs[:5]}")
err_ctr = np.array([ 
    abs(deriv_central(x0, h) - df_true(x0)) 
    for h in hs
]) 

print(f"error of central derivitive: {err_ctr[:4]}")
print(f"error of forward derivitive: {err_fwd[:4]}")

```

```python
plt.figure(figsize=(6, 4)) 
plt.loglog(hs, err_fwd, "o-", label="forward") 
plt.loglog(hs, err_ctr, "s-", label="central") 
plt.gca().invert_xaxis() 
 
plt.xlabel("h") 
plt.ylabel("absolute error") 
plt.legend() 
plt.title("Finite-Difference Error vs Step Size") 
plt.tight_layout() 
plt.show() 
```

# 1 Question: Which method gives a smaller error near its optimal h? 
### As we see from this diagram, our smaller error is between 10^-4 and 10^-6 step, and it means that our best method is central derivative method
# 2 Question: Why does the error eventually increase when h becomes extremely small? 
### Because our numpy array's dtype by default is float64, and when it becomes extremelly small our computer cant calculate the whole datas because it seems like a same datas like 1.0000000000 ~ 1.000000001
# 3 Question: Why is the central-difference method generally more accurate than the forward-difference method? 
### Because in forward derivative we do only one step to the one direction, in central der. we are doing 2 step to diff direction and by simmetric it increase our accuracy in making the our error smaller

```python
print("best h, forward:", hs[np.argmin(err_fwd)]) 
print("best h, central:", hs[np.argmin(err_ctr)]) 
 
assert abs( 
    deriv_central(x0, 1e-5) - df_true(x0) 
) < 1e-7
 
print("Task 1 OK") 
```

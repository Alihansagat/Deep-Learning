```python
import numpy as np 
 
# X: 4 examples, 2 features each 
X = np.array([ 
    [1.0, 2.0], 
    [2.0, 0.5], 
    [3.0, 1.0], 
    [4.0, 3.0], 
]) 

y = np.array([1.0, 3.0, 5.0, -2.0])
 
w = np.array([2.0, -1.0])   # shape (2,) 
b = 0.5 
 
 
def predict(X, w, b): 
    # TODO: return X @ w + b, shape should be (4,) 
    return X @ (w + b)
    
 
 
yhat = predict(X, w, b) 
print("yhat shape:", yhat.shape) 
print("yhat      :", yhat) 


#! 1st problem
res = None
summa = 0
 
for i in range(len(X)):
    summa += (y[i] - yhat[i])**2
    res = summa / len(X)
    
MSE1 = res

print(MSE1)
 
#! 2nd problem
summa = np.sum(np.array((y - yhat)**2))

MSE2 = summa / len(X)

#! 3rd problem
assert MSE1 == MSE2
print("Task 5 OK")
   
#! 4th problem
reshaped = predict(X,w,b).reshape(4,1)

summa = np.sum(np.array((y - reshaped)**2))

MSE3 = summa / len(X)
print(MSE3)

```
